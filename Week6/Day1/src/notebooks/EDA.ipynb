```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/final.csv")


df.head()
```

```python
# get number of null values in each column
df.isnull().sum()
```

```python
sns.heatmap(df.isnull()) #visualize missing values, cbar=False to hide color bar
plt.show()
```

```python
#outliner detection for all numerical columns
numerical_cols=df.select_dtypes(include='number')
for col in numerical_cols:
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python
df.select_dtypes(include='number').hist(figsize=(14, 10)) #histograms for numeric features
plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")

plt.figure(figsize=(8, 6))
sns.heatmap(
    numeric_df.corr(),
    annot=True,          # show values
    fmt=".2f",           # 2 decimal places
    cmap="coolwarm"
)
plt.title("Correlation Matrix (Numerical Features Only)")
plt.show()
```

```python
sns.countplot(x=df["income"]) #countplot for target variable to check class balance
plt.title("Income Distribution")
plt.show()
```
